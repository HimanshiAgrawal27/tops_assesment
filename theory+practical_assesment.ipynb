````python
'''Task 2: Restaurant Performance Analyser (Pandas)
Build a Pandas program that creates a food delivery dataset and produces a filtered, ranked
summary of restaurant performance using groupby aggregation.
Construct a DataFrame with at least 40 rows containing columns: restaurant_name (5 unique
names), city (3 unique cities), order_value, delivery_time_mins, rating (1.0–5.0), and
cuisine_type (3 types).
Use groupby() and agg() in a single method chain to compute: mean order value, mean
delivery time, and mean rating — grouped by restaurant_name.
Filter the grouped result to retain only restaurants with a mean rating above 4.0 and a mean
delivery time below 35 minutes.
Sort the final output by mean order value in descending order, reset the index, and print the
complete result with column headers clearly visible.'''
````

````python
import pandas as pd
import numpy as np

# Set random seed for reproducible results
np.random.seed(42)

# 1. Construct the DataFrame (40 rows)
num_rows = 40
restaurants = ['Spice Symphony', 'Burger Bonanza', 'Sushi Supreme', 'Pizza Paradise', 'Taco Town']
cities = ['Vadodara', 'Mumbai', 'Pune']
cuisines = ['Indian', 'American', 'Japanese']

data = {
    'restaurant_name': np.random.choice(restaurants, num_rows),
    'city': np.random.choice(cities, num_rows),
    'order_value': np.random.uniform(15.0, 150.0, num_rows).round(2),
    # Skewing data slightly so means hover around passing values
    'delivery_time_mins': np.random.randint(15, 45, num_rows),
    'rating': np.random.uniform(3.5, 5.0, num_rows).round(1),
    'cuisine_type': np.random.choice(cuisines, num_rows)
}

df = pd.DataFrame(data)

# Display a preview of the raw dataset
print("--- Raw Dataset (First 5 Rows) ---")
print(df.head(), "\n")

# 2. Process data in a single method chain
final_df = (
    df.groupby('restaurant_name')
    .agg(
        mean_order_value=('order_value', 'mean'),
        mean_delivery_time=('delivery_time_mins', 'mean'),
        mean_rating=('rating', 'mean')
    )
    # Use .loc with a lambda function to filter within the continuous chain
    .loc[lambda x: (x['mean_rating'] > 4.0) & (x['mean_delivery_time'] < 35)]
    .sort_values(by='mean_order_value', ascending=False)
    .reset_index()
)

# 3. Print the final filtered and ranked output
print("--- Filtered & Ranked Restaurant Performance ---")
print(final_df.to_string(index=False))
````

````python
'''Task 3: Monthly Order Trend Dashboard (Matplotlib)
Build a Matplotlib figure with three subplots that together form a visual performance report for
a food delivery platform across a 12-month period.
Generate synthetic monthly data using NumPy: 12 months of total orders (random integers
1,000–5,000), monthly revenue (orders × random average order value between Rs 200–Rs
400), and average delivery time (normally distributed, mean=28, std=4).
Subplot 1 — Line chart: Plot total orders per month with circular markers; annotate each data
point with its value displayed just above the marker.
Subplot 2 — Bar chart: Plot monthly revenue; colour each bar green if revenue exceeds Rs
8,00,000 and red otherwise, and draw a horizontal dashed line at the Rs 8,00,000 threshold.
Subplot 3 — Histogram: Plot 500 simulated delivery times (use the generated mean and std)
with 15 bins; add a vertical dashed line at the sample mean labelled 'Mean'.
Apply a figure title, individual subplot titles, axis labels on all subplots, and plt.tight_layout();
save the figure as food_delivery_dashboard.png at DPI 150.'''
````

````python
import matplotlib.pyplot as plt
import numpy as np

# Set random seed for reproducibility
np.random.seed(42)

# --- 1. Data Generation ---
months = np.arange(1, 13)
month_labels = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun',
                'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']

# 12 months of total orders (random integers 1,000–5,000)
total_orders = np.random.randint(1000, 5001, size=12)

# Average order value between Rs 200 - 400
avg_order_value = np.random.uniform(200, 400, size=12)

# Monthly revenue (orders × average order value)
monthly_revenue = total_orders * avg_order_value

# 500 simulated delivery times (mean=28, std=4)
delivery_mean_param = 28
delivery_std_param = 4
simulated_deliveries = np.random.normal(loc=delivery_mean_param,
                                        scale=delivery_std_param,
                                        size=500)

# --- 2. Figure and Subplots Setup ---
fig, axes = plt.subplots(nrows=3, ncols=1, figsize=(10, 15))
fig.suptitle('Food Delivery Platform - 12 Month Performance Dashboard',
             fontsize=16, fontweight='bold')

# --- Subplot 1: Line Chart (Total Orders) ---
ax1 = axes[0]
ax1.plot(months, total_orders, marker='o', linestyle='-', color='#1f77b4', markersize=8)
ax1.set_title('Total Orders per Month')
ax1.set_xlabel('Month')
ax1.set_ylabel('Number of Orders')
ax1.set_xticks(months)
ax1.set_xticklabels(month_labels)

# Annotate each data point just above the marker
for i, txt in enumerate(total_orders):
    ax1.annotate(str(txt),
                 (months[i], total_orders[i]),
                 textcoords="offset points",
                 xytext=(0, 10),
                 ha='center')

# --- Subplot 2: Bar Chart (Monthly Revenue) ---
ax2 = axes[1]
revenue_threshold = 800000

# Colour bars green if > 8,00,000, else red
bar_colors = ['green' if rev > revenue_threshold else 'red' for rev in monthly_revenue]
ax2.bar(months, monthly_revenue, color=bar_colors, edgecolor='black')

# Draw horizontal dashed line at threshold
ax2.axhline(y=revenue_threshold, color='black', linestyle='--', linewidth=2,
            label=f'Target Threshold (Rs {revenue_threshold:,})')

ax2.set_title('Monthly Revenue')
ax2.set_xlabel('Month')
ax2.set_ylabel('Revenue (Rs)')
ax2.set_xticks(months)
ax2.set_xticklabels(month_labels)
ax2.legend()

# --- Subplot 3: Histogram (Delivery Times) ---
ax3 = axes[2]
ax3.hist(simulated_deliveries, bins=15, color='skyblue', edgecolor='black')

# Add vertical dashed line at sample mean
sample_mean = np.mean(simulated_deliveries)
ax3.axvline(x=sample_mean, color='red', linestyle='--', linewidth=2,
            label=f'Mean: {sample_mean:.1f} mins')

ax3.set_title('Distribution of Delivery Times (500 Sampled Orders)')
ax3.set_xlabel('Delivery Time (minutes)')
ax3.set_ylabel('Frequency')
ax3.legend()

# --- 3. Final Formatting and Saving ---
plt.tight_layout()

# Save the figure as specified
plt.savefig('food_delivery_dashboard.png', dpi=150)

# Display the plot
plt.show()
````

````python
'''Task 4: Full EDA Pipeline — Food Delivery Insights (NumPy
Build an end-to-end exploratory analysis pipeline that generates, cleans, enriches, and
visualises a synthetic food delivery dataset using all four modules in sequence.
Use NumPy (seed=7) to generate a 200-row dataset with columns: order_value (uniform Rs
100–Rs 800), distance_km (uniform 1–20), delivery_time_mins (normal ?=30, ?=8), rating
(uniform 1.0–5.0), and discount_pct (uniform 0–30); load it into a Pandas DataFrame.
Artificially introduce 5% null values into the delivery_time_mins and rating columns using
np.random.choice(); then fill all nulls with the respective column medians.
Add a derived column: delivery_speed_kmph = distance_km / (delivery_time_mins / 60);
classify each row into a speed_band category ('Fast', 'Normal', 'Slow') using pd.qcut() on
delivery_speed_kmph with 3 equal-frequency bins.
Produce a Seaborn heatmap of the Pearson correlation matrix for all numeric columns with
annotated values (annot=True) and a diverging colour palette; save as
correlation_heatmap.png.
Produce a Seaborn pairplot of order_value, distance_km, delivery_time_mins, and rating,
coloured by speed_band; save as pairplot.png. Both charts must be saved at DPI >= 150 using plt.savefig().'''
````

````python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# ==========================================
# 1. Data Generation (NumPy & Pandas)
# ==========================================
# Set seed for reproducibility
np.random.seed(7)
n_rows = 200

# Generate synthetic arrays
order_value = np.random.uniform(100, 800, n_rows)
distance_km = np.random.uniform(1, 20, n_rows)
delivery_time_mins = np.random.normal(loc=30, scale=8, size=n_rows)
rating = np.random.uniform(1.0, 5.0, n_rows)
discount_pct = np.random.uniform(0, 30, n_rows)

# Load into Pandas DataFrame
df = pd.DataFrame({
    'order_value': order_value,
    'distance_km': distance_km,
    'delivery_time_mins': delivery_time_mins,
    'rating': rating,
    'discount_pct': discount_pct
})

# ==========================================
# 2. Data Cleaning
# ==========================================
# Calculate 5% of the total rows for null injection
num_nulls = int(0.05 * n_rows)

# Randomly select indices using np.random.choice
null_idx_time = np.random.choice(n_rows, num_nulls, replace=False)
null_idx_rating = np.random.choice(n_rows, num_nulls, replace=False)

# Inject nulls
df.loc[null_idx_time, 'delivery_time_mins'] = np.nan
df.loc[null_idx_rating, 'rating'] = np.nan

# Fill nulls with respective column medians
df['delivery_time_mins'] = df['delivery_time_mins'].fillna(df['delivery_time_mins'].median())
df['rating'] = df['rating'].fillna(df['rating'].median())

# ==========================================
# 3. Data Enrichment
# ==========================================
# Add derived column for speed (km/h)
df['delivery_speed_kmph'] = df['distance_km'] / (df['delivery_time_mins'] / 60)

# Classify into speed_band category using pd.qcut (3 equal-frequency bins)
# qcut orders values from lowest to highest, so our labels go Slow -> Normal -> Fast
df['speed_band'] = pd.qcut(
    df['delivery_speed_kmph'],
    q=3,
    labels=['Slow', 'Normal', 'Fast']
)

# ==========================================
# 4. Visualisation (Seaborn & Matplotlib)
# ==========================================
# Filter to numeric columns only for correlation
numeric_df = df.select_dtypes(include=[np.number])
corr_matrix = numeric_df.corr(method='pearson')

# Plot 1: Correlation Heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(
    corr_matrix,
    annot=True,
    cmap='vlag',  # Diverging colour palette
    fmt=".2f",
    center=0
)
plt.title('Pearson Correlation Matrix of Delivery Metrics', pad=15)
plt.tight_layout()
plt.savefig('correlation_heatmap.png', dpi=150)
plt.close() # Clear the figure to prevent overlap

# Plot 2: Pairplot
# Specify the columns and use speed_band for hue
pairplot_fig = sns.pairplot(
    df,
    vars=['order_value', 'distance_km', 'delivery_time_mins', 'rating'],
    hue='speed_band',
    palette='Set1',
    plot_kws={'alpha': 0.7}
)
pairplot_fig.fig.suptitle('Pairplot of Delivery Metrics by Speed Band', y=1.02)

# Save pairplot
pairplot_fig.savefig('pairplot.png', dpi=150)
print("Data processing complete. Saved 'correlation_heatmap.png' and 'pairplot.png'.")
````

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# ---------------------------------------------------------
# 1. SETUP: Make simple data and fix missing values
# ---------------------------------------------------------
def setup_data():
    # Create 200 rows of random data
    df = pd.DataFrame({
        'Restaurant': np.random.choice(['KFC', 'Subway', 'Dominos'], 200),
        'Order_Value': np.random.randint(100, 800, 200),
        'Delivery_Time': np.random.randint(15, 60, 200),
        'Rating': np.random.uniform(2.0, 5.0, 200)
    })

    # Fill any missing values with the median (middle) number
    df.fillna(df.median(numeric_only=True), inplace=True)
    return df

# ---------------------------------------------------------
# 2. MENU FUNCTIONS (One simple trick per function)
# ---------------------------------------------------------
def option_1_numpy(df):
    # Use NumPy to find the highest and lowest delivery times
    times = df['Delivery_Time'].to_numpy()
    print(f"\nFastest Delivery: {np.min(times)} mins")
    print(f"Slowest Delivery: {np.max(times)} mins")

def option_2_matplotlib(df):
    # Make a simple bar chart of Ratings
    plt.hist(df['Rating'])
    plt.title("Rating Distribution")
    plt.savefig('easy_histogram.png')
    plt.close()
    print("\nSaved chart as 'easy_histogram.png'")

def option_3_seaborn(df):
    # Make a simple heatmap of how numbers relate to each other
    numeric_data = df[['Order_Value', 'Delivery_Time', 'Rating']]
    sns.heatmap(numeric_data.corr(), annot=True)
    plt.savefig('easy_heatmap.png')
    plt.close()
    print("\nSaved chart as 'easy_heatmap.png'")

def option_4_pandas(df):
    # Group by restaurant to find the average rating
    averages = df.groupby('Restaurant')['Rating'].mean()
    print("\nAverage Ratings:")
    print(averages)

# ---------------------------------------------------------
# 3. MAIN MENU LOOP
# ---------------------------------------------------------
# Get our data ready
my_data = setup_data()

# Keep asking the user what they want to do until they type '5'
while True:
    print("\n--- MENU ---")
    print("1. Min/Max Times (NumPy)")
    print("2. Save Chart (Matplotlib)")
    print("3. Save Heatmap (Seaborn)")
    print("4. Restaurant Averages (Pandas)")
    print("5. Exit")

    choice = input("Pick a number (1-5): ")

    if choice == '1':
        option_1_numpy(my_data)
    elif choice == '2':
        option_2_matplotlib(my_data)
    elif choice == '3':
        option_3_seaborn(my_data)
    elif choice == '4':
        option_4_pandas(my_data)
    elif choice == '5':
        print("Goodbye!")
        break
    else:
        print("Invalid choice, try again.")
````

````python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# ==========================================
# 1. GENERATE SYNTHETIC DATA
# ==========================================
np.random.seed(42) # Ensures the random data is reproducible
num_rows = 250

# Generate at least 5 numeric columns using NumPy
delivery_time = np.random.normal(loc=35, scale=8, size=num_rows)
order_value = np.random.normal(loc=50, scale=20, size=num_rows)
delivery_fee = np.random.normal(loc=5, scale=1.5, size=num_rows)
distance_miles = np.random.uniform(low=0.5, high=10.5, size=num_rows)
customer_rating = np.random.uniform(low=1.0, high=5.0, size=num_rows)

# Generate 1 categorical column
cuisines = ['Italian', 'Mexican', 'Indian', 'Chinese', 'American']
cuisine_type = np.random.choice(cuisines, size=num_rows)

# Load into a Pandas DataFrame
df = pd.DataFrame({
    'Delivery_Time_min': delivery_time,
    'Order_Value_USD': order_value,
    'Delivery_Fee_USD': delivery_fee,
    'Distance_Miles': distance_miles,
    'Customer_Rating': customer_rating,
    'Cuisine_Type': cuisine_type
})

# ==========================================
# 2. SEABORN PAIRPLOT
# ==========================================
# Create pairplot colored by the categorical column
pairplot_fig = sns.pairplot(df, hue='Cuisine_Type', dropna=True)

# Export as PNG
pairplot_fig.savefig('pairplot_delivery_metrics.png', dpi=150, bbox_inches='tight')
plt.close() # Clears the figure from memory

# ==========================================
# 3 & 4. PEARSON CORRELATION MATRIX & HEATMAP
# ==========================================
# Calculate Pearson correlation
# (numeric_only=True is critical here to avoid errors)
corr_matrix = df.corr(method='pearson', numeric_only=True)

# Set up the matplotlib figure
plt.figure(figsize=(10, 8))

# Create heatmap with annotations and a diverging palette ('coolwarm')
sns.heatmap(corr_matrix,
            annot=True,
            fmt=".2f",
            cmap='coolwarm',
            vmin=-1, vmax=1, # Sets absolute scale for diverging colors
            linewidths=0.5)

plt.title('Pearson Correlation of Food Delivery Metrics')

# Export as separate PNG at DPI 200
plt.savefig('correlation_heatmap_report.png', dpi=200, bbox_inches='tight')
plt.close()

print("Dataset generated and all visualisations successfully saved as PNG files.")
````
